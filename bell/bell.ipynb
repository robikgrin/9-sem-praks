```python
import numpy as np
import pandas as pd
```

```python
def coin_mean(cc1, cc2, cc3 ,cc4):
    return (cc1 - cc2 - cc3 + cc4)/(cc1 + cc2 + cc3 +cc4)

def coin_std(cc1, cc2, cc3 ,cc4, std1, std2, std3, std4):
    summa = cc1 + cc2 + cc3 + cc4
    S1 = (1/summa - cc1/summa**2) * std1
    S2 = (1/summa - cc2/summa**2) * std2
    S3 = (1/summa - cc3/summa**2) * std3
    S4 = (1/summa - cc4/summa**2) * std4
    return np.sqrt(S1**2 + S2**3 + S3**2 + S4**2)
```

```python
path = '/Users/User/Desktop/VS code/prak/9 sem praks/bell/check_bell_+'
to_path = ['/Bell_3_a_11_5_b_0.txt', '/Bell_3_a_11_5_b_22_5.txt',
           '/Bell_3_a_34_b_0.txt', '/Bell_3_a_34_b_22_5.txt']

A1B1, A1B2, A2B1, A2B2 = 0, 0, 0, 0  
std_A1B1, std_A1B2, std_A2B1, std_A2B2 = 0, 0, 0, 0
for _ in range(4):
    exp = pd.read_csv(path + to_path[_], sep='\t')
    res = exp.mean()
    pogr = exp.std()
    if _ == 0:
        A1B1 = coin_mean(res['Coincidence Count 1'], res['Coincidence Count 2'], res['Coincidence Count 3'], res['Coincidence Count 4'])
        std_A1B1 = coin_std(res['Coincidence Count 1'], res['Coincidence Count 2'], res['Coincidence Count 3'], res['Coincidence Count 4'],
                            pogr['Coincidence Count 1'], pogr['Coincidence Count 2'], pogr['Coincidence Count 3'], pogr['Coincidence Count 4'])

    elif _ == 1:
        A1B2 = coin_mean(res['Coincidence Count 1'], res['Coincidence Count 2'], res['Coincidence Count 3'], res['Coincidence Count 4'])
        std_A1B2 = coin_std(res['Coincidence Count 1'], res['Coincidence Count 2'], res['Coincidence Count 3'], res['Coincidence Count 4'],
                            pogr['Coincidence Count 1'], pogr['Coincidence Count 2'], pogr['Coincidence Count 3'], pogr['Coincidence Count 4'])
    elif _ == 2:
        A2B1 = coin_mean(res['Coincidence Count 1'], res['Coincidence Count 2'], res['Coincidence Count 3'], res['Coincidence Count 4'])
        std_A2B1 = coin_std(res['Coincidence Count 1'], res['Coincidence Count 2'], res['Coincidence Count 3'], res['Coincidence Count 4'],
                            pogr['Coincidence Count 1'], pogr['Coincidence Count 2'], pogr['Coincidence Count 3'], pogr['Coincidence Count 4'])
    else:
        A2B2 = coin_mean(res['Coincidence Count 1'], res['Coincidence Count 2'], res['Coincidence Count 3'], res['Coincidence Count 4'])
        std_A2B2 = coin_std(res['Coincidence Count 1'], res['Coincidence Count 2'], res['Coincidence Count 3'], res['Coincidence Count 4'],
                            pogr['Coincidence Count 1'], pogr['Coincidence Count 2'], pogr['Coincidence Count 3'], pogr['Coincidence Count 4'])

print(f'S = {A1B1 - A2B1 + A1B2 + A2B2} +- {np.sqrt(std_A1B1**2 + std_A1B2**2 + std_A2B1**2 + std_A2B2**2)}')
```

```python
import matplotlib.pyplot as plt

path = '/Users/User/Desktop/VS code/prak/9 sem praks/bell/draw_cor_1'
to_path = ['/Bell_1_angle_0.txt', '/Bell_1_angle_10.txt', '/Bell_1_angle_15.txt', '/Bell_1_angle_20.txt',
           '/Bell_1_angle_22_5.txt', '/Bell_1_angle_25.txt', '/Bell_1_angle_30.txt', '/Bell_1_angle_35.txt',
           '/Bell_1_angle_40.txt', '/Bell_1_angle_45.txt', '/Bell_1_angle_50.txt', '/Bell_1_angle_55.txt',
           '/Bell_1_angle_60.txt']

cc1, cc2 ,cc3 ,cc4 = np.array([]), np.array([]), np.array([]), np.array([])
std_cc1, std_cc2, std_cc3, std_cc4 = np.array([]), np.array([]), np.array([]), np.array([])
angle = [0, 10, 15, 20, 22.5, 25, 30, 35, 40, 45, 50, 55, 60]
for _ in range(len(to_path)):
    exp = pd.read_csv(path + to_path[_], sep='\t')
    res = exp.mean()
    pogr = exp.std()
    cc1 = np.append(cc1 , (res['Coincidence Count 1']))
    cc2 = np.append(cc2 , (res['Coincidence Count 2']))
    cc3 = np.append(cc3 , (res['Coincidence Count 3']))
    cc4 = np.append(cc4 , (res['Coincidence Count 4']))

    std_cc1 = np.append(std_cc1, pogr['Coincidence Count 1'])
    std_cc2 = np.append(std_cc2, pogr['Coincidence Count 2'])
    std_cc3 = np.append(std_cc3, pogr['Coincidence Count 3'])
    std_cc4 = np.append(std_cc4, pogr['Coincidence Count 4'])


fig = plt.figure(dpi=400)
ax = fig.gca()


plt.scatter(angle, cc1, std_cc1, color='black')
plt.fill_between(angle, cc1 - std_cc1, cc1 + std_cc1, color='black', alpha=0.4,label=r'\ket{HH}')

plt.scatter(angle, cc2, std_cc2, color='blue')
plt.fill_between(angle, cc2 - std_cc2, cc2 + std_cc2, color='blue', alpha=0.4, label=r'\ket{HV}')

plt.scatter(angle, cc3, std_cc3, color='green')
plt.fill_between(angle, cc3 - std_cc3, cc3 + std_cc3, color='green', alpha=0.4, label=r'\ket{VH}')

plt.scatter(angle, cc4, std_cc4, color='red')
plt.fill_between(angle, cc4 - std_cc4, cc4 + std_cc4, color='red', alpha=0.4, label=r'\ket{VV}')

plt.xlabel('Угол')
plt.ylabel('Значение схемы совпадений')

plt.title(r'График для схемы совпадений при $\phi = 0$ у HWP(\phi)$')
plt.grid(zorder=0)
plt.show()
```
